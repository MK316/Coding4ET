## I. What is a dataframe?

A dataframe is a table-like data structure with rows and columns.

- Row: one learning item
- Column: one type of information
- Cell: one specific value
- CSV file: a common file format for storing dataframe data

데이터프레임은 엑셀 표와 비슷하지만, 앱이나 파이썬 코드가 읽기 쉽도록 정리된 자료 구조이다.  
영어교육 앱에서는 단어, 뜻, 예문, 정답, 난이도, 단원 정보 등을 데이터프레임으로 정리할 수 있다.

## A. Wide format

In **wide format**, each variable or measurement is stored in a separate column.

Wide format is usually easy for humans to read.

### Example: Wide format

| Student | Reading | Listening | Speaking | Writing |
|---|---:|---:|---:|---:|
| S1 | 85 | 90 | 80 | 88 |
| S2 | 78 | 82 | 75 | 80 |
| S3 | 92 | 88 | 86 | 91 |

### Key idea

Each student has one row.  
Each skill has its own column.

한 학생이 한 행에 있고, 읽기, 듣기, 말하기, 쓰기 점수가 각각 다른 열에 들어 있다.  
사람이 보기에는 직관적이지만, 특정 분석이나 그래프를 그릴 때는 불편할 수 있다.

## B. Narrow format

In **narrow format**, each observation is stored in a separate row.

Narrow format is often better for analysis, grouping, filtering, and visualization.

### Example: Narrow format

| Student | Skill | Score |
|---|---|---:|
| S1 | Reading | 85 |
| S1 | Listening | 90 |
| S1 | Speaking | 80 |
| S1 | Writing | 88 |
| S2 | Reading | 78 |
| S2 | Listening | 82 |
| S2 | Speaking | 75 |
| S2 | Writing | 80 |
| S3 | Reading | 92 |
| S3 | Listening | 88 |
| S3 | Speaking | 86 |
| S3 | Writing | 91 |

### Key idea

Each row represents one observation.  
The skill type is stored in one column, and the score is stored in another column.

좁은 형태에서는 한 학생의 여러 점수가 여러 행으로 나뉜다.  
`Skill` 열은 어떤 영역의 점수인지 알려주고, `Score` 열은 실제 점수를 담는다.

## 3. Main difference

| Point | Wide format | Narrow format |
|---|---|---|
| Basic shape | Many columns | Many rows |
| Human readability | Easy to read | Less compact |
| Data analysis | Sometimes less convenient | Often more convenient |
| Graphing | May need reshaping | Easier for many plots |
| Grouping | Less flexible | More flexible |
| Example use | Grade sheet, summary table | Survey analysis, visualization, repeated measures |

### Simple explanation

Wide format is good for viewing.  
Narrow format is good for analyzing.

Wide format은 사람이 표를 한눈에 볼 때 편리하다.  
Narrow format은 코딩으로 분석하거나 그래프를 그릴 때 더 유연하다.

## ❄️ Data types in a simple dataframe

In this example, we create a small dataframe with different types of data.

### Basic data types

- **Integer**: a whole number, such as `1`, `2`, or `3`
- **String**: text data, such as `"apple"` or `"school"`
- **List**: an ordered collection of values, such as `[1, 2, 3]`
- **Dictionary**: a collection of key-value pairs, such as `"Word": ["apple", "school", "borrow"]`
- **DataFrame**: a table-like data structure made of rows and columns


```python
import pandas as pd

data = {
    "ID": [1, 2, 3],
    "Word": ["apple", "school", "borrow"],
    "Meaning": ["사과", "학교", "빌리다"],
    "Level": [1, 1, 2]
}

df = pd.DataFrame(data)
df
```


### In the above example

- `ID` contains integer data.
- `Word` contains English string data.
- `Meaning` contains Korean string data.
- `Level` contains integer data.
- Each column is stored as a list.
- The whole dataset is first written as a dictionary.
- `pd.DataFrame(data)` changes the dictionary into a dataframe.

데이터프레임을 만들기 위해서는 먼저 자료를 일정한 구조로 정리해야 한다.  
여기서는 각 열의 이름을 dictionary의 key로 만들고, 각 열에 들어갈 값을 list로 정리한다.  
그 다음 `pd.DataFrame(data)`를 사용하면 행과 열이 있는 표 형태의 데이터프레임으로 바뀐다.

## II. Why does dataframe matter?

A learning app can only use the information that exists in the dataframe. (📌 controlled data)

For example:

- If we want to filter words by level, we need a `Level` column.
- If we want to show example sentences, we need an `Example` column.
- If we want to check answers, we need an `Answer` column.
- If we want to give feedback, we need a `Feedback` column.

앱 기능은 대부분 데이터프레임의 열에서 시작된다. 따라서 어떤 앱을 만들고 싶은지 생각한 뒤, 필요한 열을 먼저 설계해야 한다.

```python
df["Level"]
```

```python
df[df["Level"] == 1]
```

### ❄️ Key idea

If a feature is not represented in the data, the app cannot use it.

데이터에 없는 정보는 앱에서 사용할 수 없다.  
예를 들어 난이도별 학습 기능을 만들고 싶다면 반드시 `Level` 열이 있어야 한다.

## III. Good dataframe vs. bad dataframe

A good dataframe has a regular structure.

+ Each row should represent one item.  
+ Each column should represent one type of information.

```python
bad_data = {
    "Word List": ["Chapter 1", "apple", "school", "Chapter 2", "borrow"],
    "Meaning": ["", "사과", "학교", "", "빌리다"],
    "Level": ["", "Easy", "Easy", "", "Medium"]
}

bad_df = pd.DataFrame(bad_data)
bad_df
```

### ❄️ Problem of the above dataframe

This dataframe is easy for humans to read, but difficult for an app to process.

Problems:

- Chapter information appears inside the data rows.
- Some rows are not real learning items.
- Blank cells may cause errors.
- The structure is not consistent.

엑셀에서는 보기 좋게 보일 수 있지만, 앱은 중간 제목이나 빈칸을 잘 해석하지 못한다.  
`Chapter 1`, `Chapter 2` 같은 정보는 행 중간에 넣지 말고 별도의 열로 만들어야 한다.

```python
good_data = {
    "Chapter": [1, 1, 2],
    "Word": ["apple", "school", "borrow"],
    "Meaning": ["사과", "학교", "빌리다"],
    "Level": ["Easy", "Easy", "Medium"]
}

good_df = pd.DataFrame(good_data)
good_df
```

### ❄️ Better structure

In this dataframe:

- Each row is one word.
- Chapter information is stored in the `Chapter` column.
- There are no title rows inside the data.
- The app can filter, search, and display the data more easily.

**좋은 데이터프레임은 사람이 보기에만 좋은 표가 아니라, 앱이 읽기 좋은 표이다**.

Now we will practice basic dataframe operations.

오늘은 복잡한 코딩보다 학습앱 제작에 자주 필요한 기본 기능만 연습한다.

`quiz_df.iloc[0]` means “select the first row from the dataframe.”

In Python, counting starts from 0.  
So, `0` means the first row.

선택된 첫 번째 행은 `item`이라는 변수에 저장된다.  
이후에는 `item["Word"]`, `item["Answer"]`처럼 그 행 안에 있는 특정 열의 값을 꺼내 사용할 수 있다.

```python
#

item = quiz_df.iloc[0]
item
```

```python
item["Word"]
```

```python
random_item = quiz_df.sample(1).iloc[0]

print("Word:", random_item["Word"])
print("Answer:", random_item["Answer"])
```

### APP example

```python
#@markdown Code to run

from gtts import gTTS
from IPython.display import Audio, display
import ipywidgets as widgets
import os

# -----------------------------
# Selected item
# -----------------------------
word = random_item["Word"]
answer = random_item["Answer"]

print("Selected word:", word)

# -----------------------------
# English TTS: word + frame sentence
# -----------------------------
english_text = f"{word}. What's the meaning of {word}, can you guess?"

tts_en = gTTS(text=english_text, lang="en")
english_audio_file = "question_audio.mp3"
tts_en.save(english_audio_file)

display(Audio(english_audio_file, autoplay=False))

# -----------------------------
# Answer button
# -----------------------------
answer_button = widgets.Button(
    description="정답확인",
    button_style="success"
)

output = widgets.Output()

def show_answer(b):
    with output:
        output.clear_output()

        korean_text = f"{word}의 뜻은 {answer}입니다."
        print(korean_text)

        tts_ko = gTTS(text=korean_text, lang="ko")
        korean_audio_file = "answer_audio.mp3"
        tts_ko.save(korean_audio_file)

        display(Audio(korean_audio_file, autoplay=True))

answer_button.on_click(show_answer)

display(answer_button, output)
```

## VII. Excel rules for app-friendly data

When preparing CSV files in Excel, follow these rules.

### Rules

1. Use the first row as column names.
2. Do not merge cells.
3. Do not insert title rows inside the data.
4. Do not leave blank rows.
5. Use one piece of information per cell.
6. Use simple column names.
7. Add an `ID` column.
8. Save as CSV UTF-8 if Korean text is included.

엑셀에서 보기 좋게 만드는 방식과 앱이 읽기 좋게 만드는 방식은 다를 수 있다.  
앱용 데이터는 장식보다 구조가 중요하다.

### Recommended column names

Good examples:

- ID
- Unit
- Word
- Meaning
- Example
- Level
- Answer
- Feedback

Avoid:

- Word Meaning
- Question/Answer
- Chapter #
- My students' answers
- 단어 뜻 예문

초보 단계에서는 영어 열 이름을 사용하는 것이 코드 오류를 줄이는 데 도움이 된다.

## VIII. Reading CSV from GitHub

In many Streamlit apps, we upload a CSV file to GitHub and read it from a raw URL.

Basic steps:

1. Prepare a CSV file.
2. Upload it to GitHub.
3. Open the file on GitHub.
4. Click `Raw`.
5. Copy the raw URL.
6. Use `pd.read_csv(url)` in Python.

깃허브에 올린 CSV 파일은 Streamlit 앱이 온라인에서 직접 불러올 수 있다.
